```python
import os
import zipfile
import shutil
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# ================== CONFIG ==================
zip_path = "/content/plantvillage.zip"   # Update this path after uploading your zip file
extract_to = "./plantvillage_dataset"   # folder where it will be extracted

# ================== EXTRACT ZIP ==================
if not os.path.exists(extract_to):
    print("Extracting ZIP file... (this may take a while)")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
    print("Extraction completed!")
else:
    print("Dataset already extracted.")

# ================== SET DATA PATH ==================
# Usually the images are inside one of these folders:
data_dir = os.path.join(extract_to, "color")        # Try this first
# data_dir = os.path.join(extract_to, "segmented")  # or try this
# data_dir = os.path.join(extract_to, "grayscale")

print("Data directory:", data_dir)
print("Available folders:", os.listdir(data_dir)[:10])  # preview classes
```
