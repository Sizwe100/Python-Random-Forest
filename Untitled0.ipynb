### Load an image from Google Drive

To load an image from your Google Drive, you first need to know its path. You can find this by navigating to the image in Google Drive, right-clicking on it, selecting 'Get link', and then copying the path. Remember that your Google Drive is mounted at `/content/drive`.

```python
import cv2
import matplotlib.pyplot as plt

# Replace 'path/to/your/image.jpg' with the actual path to your image in Google Drive
# For example: '/content/drive/MyDrive/my_images/example.jpg'
image_path = '/content/drive/MyDrive/path/to/your/image.jpg'

# Load the image
image = cv2.imread(image_path)

# Check if the image was loaded successfully
if image is not None:
    # OpenCV loads images in BGR format, convert to RGB for matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Display the image
    plt.imshow(image_rgb)
    plt.axis('off') # Hide axes ticks
    plt.title('Image from Google Drive')
    plt.show()
else:
    print(f"Error: Could not load image from {image_path}. Please check the path.")
```

```python
from ultralytics import YOLO
import os

# 1. Load your newly trained custom plant classification model
# Note: YOLO automatically saves the absolute best version as 'best.pt'
model_path = '/content/runs/classify/train-3/weights/best.pt'
custom_model = YOLO(model_path)

# 2. Pick ANY image of a plant leaf to test it out!
# REPLACE the URL below with a path to one of your test images if you prefer.
test_image = "https://ultralytics.com"

# 3. Ask the AI model to predict what it sees
print("Running prediction...")
results = custom_model(test_image)

# 4. Print out what the AI thinks the disease/state is
for result in results:
    top_class_idx = result.probs.top1
    confidence = result.probs.top1conf.item()
    class_name = result.names[top_class_idx]

    print("\n==============================")
    print(f"🌿 AI Prediction: {class_name}")
    print(f"📊 Confidence Score: {confidence * 100:.2f}%")
    print("==============================")
```
