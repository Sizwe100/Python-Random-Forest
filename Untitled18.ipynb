```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import numpy as np
```

```python
# --- 1. SET UP THE UNIFORM STYLING ---
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.size'] = 10
plt.rcParams['axes.linewidth'] = 1.0

# Define a consistent date range matching your datasets (approx. 2019 to 2024)
dates = pd.date_range(start='2019-04-01', end='2024-01-01', freq='ME')
n_points = len(dates)
```

```python
# --- 2. GENERATE REPRESENTATIVE/PLACEHOLDER DATA FOR UNIFORM PLOTTING ---
# (Replace these with your actual numpy arrays or pandas DataFrame columns)
np.random.seed(42)
def generate_cyclical_data(base, amplitude, noise, trend=0):
    # Generates a realistic cyclical environmental time series
    cycle = np.sin(2 * np.pi * np.arange(n_points) / 12)
    trend_line = np.linspace(0, trend, n_points)
    return base + (amplitude * cycle) + (np.random.randn(n_points) * noise) + trend_line

data_evi = generate_cyclical_data(0.5, 0.08, 0.04, trend=0.1)
data_et = generate_cyclical_data(3.5, 1.2, 0.2)
data_msavi_a = generate_cyclical_data(0.4, 0.06, 0.02)
data_msavi_b = generate_cyclical_data(0.28, 0.05, 0.03)
data_savi_c = generate_cyclical_data(0.42, 0.05, 0.02)
data_savi_d = generate_cyclical_data(0.31, 0.05, 0.03)
```

```python


# Group datasets and properties for sequential plotting
subplots_config = [
    {"label": "(a)", "title": "EVI_La_Chagra", "ylabel": "EVI Value", "data": data_evi, "color": "#1f77b4"},
    {"label": "(b)", "title": "EVI_UVS", "ylabel": "ET (kg/m²/8day)", "data": data_et, "color": "#2ca02c"},
    {"label": "(c)", "title": "MSAVI_La_Chagra", "ylabel": "Index Value", "data": data_msavi_a, "color": "#d62728"},
    {"label": "(d)", "title": "MSAVI_UVS", "ylabel": "Index Value", "data": data_msavi_b, "color": "#9467bd"},
    {"label": "(e)", "title": "SAVI_La_Chagra", "ylabel": "Index Value", "data": data_savi_c, "color": "#e377c2"},
    {"label": "(f)", "title": "SAVI_UVS", "ylabel": "Index Value", "data": data_savi_d, "color": "#bcbd22"}
]

# --- 3. CREATE THE 3x2 SUBPLOT GRID ---
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(14, 12), sharex=True)
axes = axes.flatten()  # Flatten into a 1D array for easy iteration

for i, config in enumerate(subplots_config):
    ax = axes[i]

    # Plot the time series line
    ax.plot(dates, config["data"], color=config["color"], linewidth=1.75, label=config["title"])

    # Uniform Panel Text Labels (a to f) in the upper-left corner
    ax.text(0.02, 0.92, config["label"], transform=ax.transAxes, fontsize=14, weight='bold', va='top')

    # Subplot Title & Axes Setup
    ax.set_title(config["title"], fontsize=11, pad=10, weight='semibold')
    ax.set_ylabel(config["ylabel"], fontsize=10)
    ax.grid(True, linestyle='--', alpha=0.5, linewidth=0.5)
    ax.tick_params(axis='both', which='both', labelsize=9)

    # Uniform Date Axis Formatting
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

# Global X-Axis Label for the bottom row panels
axes[4].set_xlabel('Date', fontsize=11, labelpad=8)
axes[5].set_xlabel('Date', fontsize=11, labelpad=8)

# --- 4. OPTIMIZE AND SAVE LAYOUT ---
plt.tight_layout()
plt.savefig('combined_environmental_time_series.png', dpi=300, bbox_inches='tight')
plt.show()
```
